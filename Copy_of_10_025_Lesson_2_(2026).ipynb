### A function can return information

A function can have a ```return``` statement.

The ```return``` statement is always the final statement in the function and terminates the function.

When the function terminates, the ```return``` statement gives back a value to the python statement that executed the function.

Thus, at statement ```(a)```,
- the function ```quadratic2``` is executed
- and because it returns a value, the value is assigned to a variable called ```result```.
- then a separate ```print``` statement displays the result on the screen.

In this way, the different tasks are neatly separated in each line i.e.
* the function ```quadratic2``` tells you the value of ```y``` for any ```x``` that you give it
* the result is stored in ```result```
* ```print(result)``` displays the result on the screen.

```python
def quadratic2(x):
    y = x*x + 5*x + 4
    return y

result = quadratic2(1) #(a)
print(result)
result = quadratic2(2) #(a)
print(result)
```

### Short-circuit evaluation is used with the boolean operators ```and``` and ```or```

The [python documentation](https://docs.python.org/3/reference/expressions.html?highlight=precedence#boolean-operations) describes how **short-circuit** evaluation works.

---
>
> 1. The expression ```x and y``` first evaluates ```x```; if ```x``` is false, its value is returned; otherwise, ```y``` is evaluated and the resulting value is returned.
>
> 2. The expression ```x or y``` first evaluates ```x```; if ```x``` is true, its value is returned; otherwise, ```y``` is evaluated and the resulting value is returned.
>
---


Consider the following statement.

```python
(2 > 1 ) or (10 < 3)
```

Python applies **short-circuit evaluation** here:
- Since ```(2 > 1)``` evaluates to ```True```, python decides that it is not necessary to evaluate ```(10 < 3)```.

- This is because evaluating ```(10 < 3)``` will not change the outcome of the expression.

### Good programming practices - boolean variables and if/else

*This section is adapted from Chapter 4 of "Introduction to Programming using Python" by Y. Daniel Liang*

The good programming practices highlighted in this section make your code easier to read. Remember, a computer program is also communication of your approach to solving a problem, both to your future self and to others.  

**Assigning Boolean Variables**

*Not so good*  Most beginner programmers would write something like this.
```python
n = ...
if ( n % 2 == 0):
    is_even = True
else:
    is_even = False
```

*Better!* Experienced programmers will simply do this, realizing that ```n % 2 == 0 ``` evaluates to ```True``` or ```False```. This makes the if/else structure described previously redundant.  Also, it is good practice to name your boolean variables with an ```is_``` to make their roles clear.

```python
n = ...
is_even = n % 2 == 0
```

**Two conditions**

*Not so good* Python allows expressions like this, which looks like what you write in mathematics.

```python
is_allowed = 1 < n < 10
```

Writing this way is not a good habit to pick up because other programming languages may not accept such expressions.

*Better!* Use the ```and``` operator.

```python
is_allowed = n > 1 and n < 10
```

**Testing conditions in if/else**

*Not so good* Since ```is_even``` is already a boolean variable, it is already ```True``` or ```False```, the condition ```is_even == True``` is redundant.

```python
is_even = ... #boolean variable
if is_even == True:   ## <-- NO GOOD
    print("Even")
else:
    print("Odd")
```

*Better!*
```python
is_even = ... #boolean variable
if is_even :  ## <-- BETTER
    print("Even")
else:
    print("Odd")
```
  

**Avoid duplicate code - Don't Repeat Yourself (DRY)**

*Not so good* - Notice that the print statement is repeated.

```python
is_even = ... #boolean variable
if is_even :   
    result = 'Even'
    print('The number is', result)
else:
    result = 'Odd'
    print('The number is', result)
```

*Better!* Apply **Don't Repeat Yourself** and move the print statement outside of the if-else.

```python
is_even = ... #boolean variable
if is_even:   
    result = 'Even'
else:
    result = 'Odd'
print('The number is', result)
```

### Recall good programming practices - test early and often

It is a good practice to implement a program in stages, rather that doing so at one go. This allows you to discover mistakes in your program earlier with a less complicated program.

Let's say I'm tasked to write a python function for this formula.

\begin{equation}
f(x,y) = \frac{3x^2y + y^2}{-(x^3 + 2xy)}
\end{equation}

First, I would implement the numerator and verify that the calculation is correct. Using a few values of x and y, I would first work out by hand the expected values, and verify that the program returns the correct value.

**First Stage**
```python
def function(x,y):
    upper = 3*x*x*y + y*y
    return upper

print(function(1,1))
```

You could also write print statements in the function.
```python
def function(x,y):
    upper = 3*x*x*y + y*y
    print(upper) ## <--- HERE
    return upper

print(function(1,1))
```

Once I'm sure that the numerator is correct, I will then implement the denominator. Similarly, using a few values of x and y, I would first work out by hand the expected values, and verify that the denominator is calculated correctly with some test cases.

Once I'm satified that all values are calculated correctly, if I wrote print statements to help me check intermediate values, I will then comment them out so that they are not executed. I could also delete them.

**Next Stage**
```python
def function(x,y):
    upper = 3*x*x*y + y*y
    ## print(upper) ## <-- now commmented out
    lower = -( x**3 + 2*x*y)
    ## print(lower) ## <-- commented out after I checked that it is correct
    return upper/lower

print(function(1,1))
```

Notice that in this series of examples, I am also using the rule of "break up long lines into shorter steps".

### Example 2.03 Factorial

Write a function ```factorial``` that takes in an integer that is 0 or larger and returns the factorial.

You have several options for this question
* You could import the ```math``` library, execute the ```factorial``` function and return the result
* You could use the formula **Stirling's Approximation**
* You could use a for-loop to calculate the factorial.  
* Any other solution you can think of


```python
print( factorial(0) )
print( factorial(3) )
```

gives


```python

````

def factorial(value):
  final_answer = 1
  for i in range(1, value + 1 ):
    final_answer = final_answer * i
  return final_answer

print( factorial(0) )
print( factorial(3) )
````
